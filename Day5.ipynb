### If-else Statements
* Write a program that takes an integer as input and checks if it's even or odd.
* Write a program that takes an age as input and determines if the person is a child, teenager, adult, or senior citizen.


```python
def even_or_odd(number):
    if number % 2 == 0:
        return "Even number"
    else:
        return "Odd number"
    
def classifying_age(age):
    if age < 15:
        return "Child"
    elif age >= 15 and age < 18:
        return "Teenager"
    elif age >= 18 and age < 65:
        return "Adult"
    else:
        return "Senior citizen"
    
if __name__ == "__main__":
    age = int(input("Enter your age: "))
    
    print(even_or_odd(age))
    print()
    print((classifying_age(age)))
```

### Nested If-else Statements
* Using nested if-else, write a program that takes three numbers as input and determines the largest among them.

```python
def largest_number(num1, num2, num3):
    larger_number = 0
    if num1 > num2 and num1 > num3:
        larger_number = num1
    elif num2 > num3 and num2 > num1:
        larger_number = num2
    else:
        larger_number = num3
    return larger_number

if __name__ == "__main__":
    num1 = 10
    num2 = 15
    num3 = 20
    
    print(f'The largest number is {largest_number(num1, num2, num3)}')
```

### For Loop
* Write a program to calculate the sum of all numbers up to the given input number.

```python
def sum_algarisms(num):
    num_str = list(str(num))
    soma = 0
    for i in num_str:
        numero = int(i)
        soma += numero
    return soma

if __name__ == "__main__":
    numero = 256
    print(sum_algarisms(numero))
```

### While Loop
* Write a program to calculate the factorial of a given number.

```python
def factorial(num):
    fat = 1
    contador = num
    while contador > 1:
        fat = fat * contador
        contador -= 1
    return fat

if __name__ == "__main__":
    num = 6
    print(factorial(num))
```
